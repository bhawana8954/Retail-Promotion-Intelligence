```python
# pick a household that actually has transactions in our sample
test_household = df_transactions['household_key'].iloc[0]
test_household
```

```python
# rerun the comparision with this corrected household
sample_transactions = df_transactions[df_transactions['household_key'] == test_household] [['household_key', 'day', 'commodity_desc', 'sales_value']]
sample_exposure = df_exposure[df_exposure['household_key'] == test_household]
print(sample_transactions.head(10))
print(sample_exposure)
```

```python
# Observation: Transactions for the selected household occur on day 1, while its campaign windows start 
# much later (days 504–551 and 587–642). Since none of the transactions fall within a campaign period, all
# observed purchases represent baseline (pre-campaign) spending.
```

```python
# define a function to check "was this day during any campaign"
def is_during_campaign(day, exposure_df):
    matches = exposure_df[(exposure_df['start_day'] <= day) & (exposure_df['end_day'] >= day)]
    return not matches.empty
```

```python
# apply this function to our sample transaction
sample_transactions['during_campaign'] = sample_transactions['day'].apply(lambda d: is_during_campaign(d, sample_exposure))
sample_transactions
```

```python
# apply this to every transaction in df_transaction (not just the test_household)
df_transactions['during_campaign'] = df_transactions.apply(lambda row: is_during_campaign(row['day'], df_exposure[df_exposure['household_key'] == row['household_key']]),
                                                        axis = 1)
df_transactions['during_campaign'].value_counts()
```

```python
# Diagnosed data filtering issue and strategized sampling refinement:
# Observation: All transactions are marked False because the TOP 1000 sample contains only early transaction days,
# before any campaigns begin. The campaign-matching logic is correct; the issue lies in the sampling method.
```

```python
# pull a sample that's actually likely to include campaign exposed transactions
query_targeted = """SELECT TOP 2000 t.*
                    FROM silver.transaction_data AS t
                    INNER JOIN silver.campaign_table AS c
                    ON t.household_key = c.household_key
                    ORDER BY t.household_key"""
df_transactions = pd.read_sql(query_targeted, connector)
df_transactions.shape
```

```python
# redo the category join and campaign-window check on this new sample
df_transactions = df_transactions.merge(df_product, on = 'product_id', how = 'left')
df_transactions['during_camapign'] = df_transactions.apply(lambda row: is_during_campaign(row['day'], df_exposure[df_exposure['household_key'] == row['household_key']]),
                                                           axis = 1)
df_transactions['during_camapign'].value_counts()
```

```python
# calculate baseline spend per household per category
baseline_spend = df_transactions[df_transactions['during_camapign'] == False].groupby(['household_key', 'commodity_desc'])['sales_value'].sum().reset_index()
baseline_spend = baseline_spend.rename(columns = {'sales_value' : 'baseline_spend'})
baseline_spend.head()
```

```python
# calculate the campaign exposed spend the same way
campaign_spend = df_transactions[df_transactions['during_camapign'] == True].groupby(['household_key', 'commodity_desc'])['sales_value'].sum().reset_index()
campaign_spend = campaign_spend.rename(columns = {'sales_value' : 'campaign_spend'})
campaign_spend.head()
```

```python
# merge baseline and campaign spend side by side
lift_df = baseline_spend.merge(campaign_spend, on = ['household_key', 'commodity_desc'], how = 'outer')
lift_df.head()
```

```python
# fill in the gaps and calculate lift
lift_df['baseline_spend'] = lift_df['baseline_spend'].fillna(0)
lift_df['campaign_spend'] = lift_df['campaign_spend'].fillna(0)
lift_df['lift'] = lift_df['campaign_spend'] - lift_df['baseline_spend']
lift_df.sort_values('lift', ascending = False).head(10)
```

```python
# Identified two issues affecting the analysis.
# Observation: First, the TOP 2000 sample contains transactions from only household 1 due to the ORDER BY clause,
# resulting in a non-representative sample. Second, comparing raw baseline and campaign spend totals
# is misleading because they may cover different time periods, making lift calculations potentially biased.
```

```python
# Lets check both
print(f'Number of unique households in transactions: {df_transactions['household_key'].nunique()}')
print(f'\nCampaign duration (days): {df_exposure['end_day'] - df_exposure['start_day']}')
```

```python
# Reconsider sample scope and normalized metrics for equitable analysis
# Observation: Verified that the current sample contains transactions from only one household due to the
# TOP 2000 + ORDER BY query. Also confirmed that campaign windows vary between 30–70 days,
# making raw baseline and campaign spend totals incomparable without normalizing for time.
```

```python
# pull a real spread of households
query_targeted = """SELECT TOP 20000 t.*
                    FROM silver.transaction_data AS t
                    WHERE t.household_key IN(
                                             SELECT DISTINCT TOP 20 household_key
                                             FROM silver.campaign_table)"""
df_transactions = pd.read_sql(query_targeted, connector)
df_transactions['household_key'].nunique()
```

```python
# check how many total days our data actually spans
df_transactions['day'].max() - df_transactions['day'].min()
```

```python
# redo category join + campaign-window flag on new sample
df_transactions = df_transactions.merge(df_product, on = 'product_id', how = 'left')
df_transactions['during_campaign'] = df_transactions.apply(lambda row: is_during_campaign(row['day'], df_exposure[df_exposure['household_key'] == row['household_key']]),
                                                           axis = 1)
df_transactions['during_campaign'].value_counts()
```

```python
# count campaign days vs baseline days per household
days_per_household = df_transactions.groupby(['household_key', 'during_campaign'])['day'].nunique().reset_index()
days_per_household = days_per_household.rename(columns = {'day' : 'num_days'})
days_per_household.head(10)
```

```python
# recalculate baseline and camapign spend on this new smaple
baseline_spend = df_transactions[df_transactions['during_campaign'] == False].groupby(['household_key', 'commodity_desc'])['sales_value'].sum().reset_index()
baseline_spend = baseline_spend.rename(columns={'sales_value': 'baseline_spend'})

campaign_spend = df_transactions[df_transactions['during_campaign'] == True].groupby(['household_key', 'commodity_desc'])['sales_value'].sum().reset_index()
campaign_spend = campaign_spend.rename(columns={'sales_value': 'campaign_spend'})
```

```python
baseline_days = days_per_household[days_per_household['during_campaign'] == False][['household_key', 'num_days']].rename(columns={'num_days': 'baseline_days'})
campaign_days = days_per_household[days_per_household['during_campaign'] == True][['household_key', 'num_days']].rename(columns={'num_days': 'campaign_days'})

baseline_spend = baseline_spend.merge(baseline_days, on='household_key', how='left')
campaign_spend = campaign_spend.merge(campaign_days, on='household_key', how='left')

baseline_spend.head()
```

```python
# calculate spend per day for both baseline and camapign
baseline_spend['baseline_spend_per_day'] = baseline_spend['baseline_spend'] / baseline_spend['baseline_days']
campaign_spend['campaign_spend_per_day'] = campaign_spend['campaign_spend'] / campaign_spend['campaign_days']

baseline_spend.head()
```

```python
# merge both per day rates together and compute the real lift
lift_df = baseline_spend.merge(campaign_spend, on=['household_key', 'commodity_desc'], how='outer')

lift_df['baseline_spend_per_day'] = lift_df['baseline_spend_per_day'].fillna(0)
lift_df['campaign_spend_per_day'] = lift_df['campaign_spend_per_day'].fillna(0)

lift_df['lift_per_day'] = lift_df['campaign_spend_per_day'] - lift_df['baseline_spend_per_day']

lift_df.sort_values('lift_per_day', ascending=False).head(10)
```

```python
# Identified data quality issues and flagged inconsistent handling across columns
# Observation: The normalized per-day metrics are more meaningful, but very small campaign periods (e.g., 1 day)
# can produce misleadingly high spend rates. Additionally, missing baseline values are handled
# inconsistently —per-day metrics are filled with 0, while the corresponding raw spend and day
# columns still contain NaN, making the results potentially misleading.
```

```python
# fix the NaN inconsistency first
lift_df['baseline_spend'] = lift_df['baseline_spend'].fillna(0)
lift_df['baseline_days'] = lift_df['baseline_days'].fillna(0)
lift_df['campaign_spend'] = lift_df['campaign_spend'].fillna(0)
lift_df['campaign_days'] = lift_df['campaign_days'].fillna(0)
lift_df.loc[1298]
```

```python
# see how common the tiny denominator problem actually is
lift_df['campaign_days'].value_counts().sort_index().head(10)
```

```python
# apply the minimum day threshold
MIN_DAYS = 5

reliable_lift = lift_df[(lift_df['baseline_days'] >= MIN_DAYS) & (lift_df['campaign_days'] >= MIN_DAYS)]
reliable_lift.shape
```

```python
# look at the top reliable lifts
reliable_lift.sort_values('lift_per_day', ascending=False).head(10)
```

```python
# also check the bottom (negative lift) as a sanity check
reliable_lift.sort_values('lift_per_day', ascending=True).head(10)
```

```python
print(f"Total pairs: {len(lift_df)}, Reliable pairs: {len(reliable_lift)}, Dropped (unreliable): {len(lift_df) - len(reliable_lift)}")
```
